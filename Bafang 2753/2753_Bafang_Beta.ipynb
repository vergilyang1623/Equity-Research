```python
import yfinance as yf

tickers = {"0050": "0050.TW", "八方雲集": "2753.TW"}

prices = yf.download(
    list(tickers.values()),
    period="5y",
    interval="1wk",
    auto_adjust=True,   # "Close" is the adjusted close
    progress=False,
)["Close"]

prices = prices.rename(columns={v: k for k, v in tickers.items()})
print(prices)
```

```python
import numpy as np
log_ret = np.log(prices / prices.shift(1)).dropna()
# Covariance matrix (sample, ddof=1)
cov_matrix = log_ret.cov()

# Covariance between the two returns
cov_both = log_ret["0050"].cov(log_ret["八方雲集"])

# Variance of 0050's return
var_0050 = log_ret["0050"].var()

beta_2753 = cov_both / var_0050
print(f"Beta of 八方雲集 (2753.TW) with respect to 0050 (0050.TW): {beta_2753:.4f}")
```
